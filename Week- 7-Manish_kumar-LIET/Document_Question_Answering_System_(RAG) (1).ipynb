```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("/content/Manish_Kumar CV.pdf")

docs = loader.load()
```

```python
pip install langchain-text-splitters
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

chunks = splitter.split_documents(docs)
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings

embedding = HuggingFaceEmbeddings(
    model_name="BAAI/bge-small-en-v1.5"
)
```

```python
from langchain_community.vectorstores import FAISS

db = FAISS.from_documents(chunks, embedding)

db.save_local("database")
```

```python
db = FAISS.load_local(
    "database",
    embedding,
    allow_dangerous_deserialization=True
)
```

```python
query = "what was his contect details?"

results = db.similarity_search(query, k=4)

for r in results:
    print(r.page_content)
```

```python
import os
from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI

# Fetch your Gemini API key from Colab secrets (Make sure you add it there first!)
GEMINI_API_KEY = userdata.get("GEMINI_API_KEY")

llm = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",
    temperature=0,
    google_api_key=GEMINI_API_KEY
)

context = "\n".join([doc.page_content for doc in results])
prompt = f"""
Answer only from the context.Format the output cleanly using Markdown bullet points and bold headers.
Do not include raw data or metadata structures.
Context:
{context}

Question:
{query}
"""
answer = llm.invoke(prompt)
if isinstance(answer.content, list) and len(answer.content) > 0:
    text_output = answer.content[0].get('text', '')
else:
    text_output = answer.content

print(text_output)
```
